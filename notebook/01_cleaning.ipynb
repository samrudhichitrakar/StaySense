```python
df = df.drop(columns=["company"])
```

```python
df["agent"] = df["agent"].fillna(0)
df["country"] = df["country"].fillna("Unknown")
df["children"] = df["children"].fillna(0)
```

```python
# Zero or negative ADR
df[df["adr"] <= 0].shape
```

```python
# Duplicates
df.duplicated().sum()
```

```python
bad_adr = df[df["adr"] <= 0]
print(bad_adr["market_segment"].value_counts())
print()
print(bad_adr["is_canceled"].value_counts())
```

```python
# Drop the 1 negative ADR row
df = df[df["adr"] >= 0]

# Flag rows with a usable rate
df["has_revenue"] = (df["adr"] > 0).astype(int)

# Core features
df["total_nights"] = df["stays_in_weekend_nights"] + df["stays_in_week_nights"]
df["total_revenue"] = df["adr"] * df["total_nights"]

# Proper arrival date
df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d"
)

# Lead time buckets
df["lead_time_bucket"] = pd.cut(
    df["lead_time"],
    bins=[-1, 30, 90, 180, 365, df["lead_time"].max()],
    labels=["0-30", "31-90", "91-180", "181-365", "365+"]
)

print(df.shape)
print(df["lead_time_bucket"].value_counts().sort_index())
```

```python
df.shape
```

```python
# Cancellation rate before cleaning (reloads the raw file just for this number)
raw = pd.read_csv("../data/hotel_bookings.csv")
rate_before = raw["is_canceled"].mean()

# Drop zero-guest bookings
df = df[(df["adults"] + df["children"] + df["babies"]) > 0]

# Drop duplicates
df = df.drop_duplicates().reset_index(drop=True)

rate_after = df["is_canceled"].mean()
print(rate_before, rate_after)
print(df.shape)
print(df["lead_time_bucket"].value_counts().sort_index())
```

```python
dupes = raw[raw.duplicated(keep="first")]
print(dupes["is_canceled"].mean())
```

```python
df.to_csv("../data/hotel_bookings_clean.csv", index=False)
```
